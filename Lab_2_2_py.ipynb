```python
#Бодлого 1. Өгсөн 2 тооны тэгш тоог хэвлэнэ үү.
a = int(input("1-р тоо: "))
b = int(input("2-р тоо: "))

for num in (a, b):
    if num % 2 == 0:
        print(f"тэгш: {num}")
```

```python
#Бодлого 2. Өгсөн 4 тооны 3-д хуваагддаг тоонуудыг нь хэвлэнэ үү.
numbers = []
for i in range(4):
    n = int(input(f"{i + 1}-р тоо: "))
    numbers.append(n)

for n in numbers:
    if n % 3 == 0:
        print(f" 3-д хуваагддаг тоонууд: {n}")
```

```python
#Бодлого 3. Өгсөн тоон үнэлгээг үсгэн үнэлгээнд шилжүүлнэ үү:
# Тоон үнэлгээ 89-с их бол А, 79-с их бол В, 69-с их бол С, 59-с их бол D бусад тохиолдолд F үсгэн үнэлгээг хэвлэнэ үү.
score = float(input("Тоон үнэлгээ: "))

if score > 89:
    grade = "A"
elif score > 79:
    grade = "B"
elif score > 69:
    grade = "C"
elif score > 59:
    grade = "D"
else:
    grade = "F"

print(f"Үсгэн үнэлгээ: {grade}")
```
