```python
#Бодлого 1. Өгөгдсөн илэрхийллийн утгыг бодит тоо X-н хувьд бодож үр дүнг таслалаас хойш  3 орны нарийвчлалтай хэвлэнэ үү.
# Y = 3x^3 - 2x^2 + 3x - 1
x = float(input("X-н утгыг оруулна уу: "))
y = 3 * x**3 - 2 * x**2 + 3 * x - 1
print(f"Y = {y:.3f}")
```

```python
#Бодлого 2. Эхлэх цаг, минут, үйл ажиллагааны үргэлжлэх хугацаа минутаар өгөгдсөн, Дуусах цаг ба минутыг олно уу.
#Жишээ нь: үйл ажиллагаа 12,00 эхэлсэн үргэлжлэх хугацаа 58 минут бол хэдэн цаг минутад үйл ажиллагаа дуусах вэ?
hour = int(input("Hour= "))
minute = int(input("Minute= "))
dura = int(input("Dura= "))

total_minutes = hour * 60 + minute + dura
end_hour = (total_minutes // 60) % 24
end_minute = total_minutes % 60

print(f"Дуусах цаг: {end_hour}:{end_minute:02d}")
```

```python
#Бодлого 3-1. Өгөгдсөн цагийг хоногт дэвшүүлнэ үү.
print("-- a) Цагийг хоногт шилжүүлэх --")
total_hours = int(input("Нийт цаг: "))  # жишээ: 24, 48, 50 гэх мэт
days = total_hours // 24
remaining_hours = total_hours % 24

if remaining_hours == 0:
    print(f"{total_hours} цаг = {days} хоног")
else:
    print(f"{total_hours} цаг = {days} хоног {remaining_hours} цаг")
```

```python
#Бодлого 3-2. Тэгш өнцөгт гурвалжны  тал катетууд 5 болон 8 см бол талбайг олно уу.
print("\n-- b) Тэгш өнцөгт гурвалжны талбай --")
a = float(input("Нэг катет (см): "))  # жишээ өгөгдөл: 5
b = float(input("Нөгөө катет (см): "))  # жишээ өгөгдөл: 8
area = 0.5 * a * b
print(f"Талбай = {area:.3f} см²")
```

```python
#Бодлого 4. А тоог В зэрэгт дэвшүүлнэ үү.
a = float(input("A: "))
b = float(input("B: "))
result = a ** b
print(f"A^B = {result}")
```

```python
#Бодлого 5. Өгөгдсөн бодит утгыг градуст шилжүүлсэн Х-н хувьд илэрхийллийг бодно уу. Y = cos(x) / sin(x)
import math

x_deg = float(input("X-н утгыг градусаар оруулна уу: "))
x_rad = math.radians(x_deg)
y = math.cos(x_rad) / math.sin(x_rad)
print(f"Y = {y:.3f}")
```
